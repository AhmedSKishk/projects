## Dataset: IMDB Top 1000 Movies

| Column | Description |
|---|---|
| `Poster_Link` | URL to the movie's poster image |
| `Series_Title` | Title of the movie |
| `Released_Year` | Year the movie was released |
| `Certificate` | Age/content rating (e.g., U, A, PG-13, R) |

| `Genre` | One or more genres, comma-separated (e.g., "Crime, Drama") |
| `IMDB_Rating` | IMDB user rating out of 10 |
| `Overview` | Short plot summary |
| `Meta_score` | Metacritic score out of 100 |
| `Director` | Movie director |
| `Star1` – `Star4` | Lead actors, in billing order |
| `No_of_Votes` | Number of IMDB user votes the rating is based on |
| `Gross` | Box office gross revenue (USD, comma-formatted string) |

# - Data Cleaning

### - Drop unnecessary Columns

```python
# Poster_Link, 	Overview >> not needed for analysis

df.drop(['Poster_Link', 'Overview'], axis= 1, inplace= True)
```

### - Investigate `Drishyam` column Series_Title

```python
for col in categ_cols:
    print(col)
    print(df[col].nunique())
    print('---------------')
    print(df[col].unique())
    print('-' * 50)
```

### - `Certificate` cleaning
    - Passed - Approved >> kept
    - UA - U/A >> merge

```python
# Passed - Approved >> kept

df['Certificate'].value_counts(dropna=False)
```

```python
# UA - U/A >> merge

df['Certificate'].replace({'U/A' : 'UA'}, inplace= True)
df['Certificate'].value_counts(dropna=False)
```

### - `Released_Year` cleaning
    - 'PG' to 'year= 1995'

```python
# one row (Apollo 13) has relased year 'PG'

not_year = ~df['Released_Year'].str.isnumeric()
df[not_year]
```

```python
# replace ('PG')'1995'

df.loc[not_year, 'Released_Year'] = '1995'
```

```python
# for check

df.loc[not_year, 'Released_Year']
```

```python
# for check

df[df['Released_Year'] == 'PG']
```

```python
# convert `Released_Year` from object to int 

df['Released_Year'] = df['Released_Year'].astype(int)

df['Released_Year'].info()          # for check
```

### - `Runtime` cleaning
    - remove (' min')

```python
# remove (' min') and convert minutes from object to int

df['Runtime'] = df['Runtime'].str.replace(' min', '').astype(int)

df['Runtime'].info()        # for check
```

### - `Gross` cleaning
    - remove (',')

```python
# remove (',') and convert thousands from object to float

df['Gross'] = df['Gross'].str.replace(',', '').astype(float)

df['Gross'].info()      # for check
```

### - Missing Values
    - `Gross` (16.9%), `Meta_score` (15.7%) and `Certificate` (10.1%) are missing values, 
    left as `NaN` rather than filled, 
    because any filled number would be invented data.

# - Data Analysis

### -  Which movie have the Heighst Revenue?

```python
# Star Wars: Episode VII

df[df['Gross'] == df['Gross'].max()][['Series_Title','Released_Year', 'Gross']]
```

```python
df.nlargest(3, 'Gross')[['Series_Title','Released_Year', 'Gross']]
```

- (Star Wars: Episode VII – The Force Awakens), earned about ($936.7M) is the most, 
followed by (Avengers: Endgame) earned about ($858.4M) and (Avatar) earned about ($760.5M).


### - What is the Average Gross Revenue for movies of the 2000's (2000 --> 2010)

```python
# Filter data first
filter_data = df[df['Released_Year'].between(2000, 2010)]

# Movies released between 2000-2010
print(f'Movies released between 2000-2010 : {len(filter_data)}')

# Average gross revenue
print('Average gross revenue:', round(filter_data['Gross'].mean(), 2))
```

- about **$73.7M** per movie, calculated on the 260 movies in this range.

### - What is the Average Run time per Director ?

```python
# Longest average
print('Longest average runtime per director')
display(df.groupby('Director')['Runtime'].mean().sort_values(ascending= False).head())

# Shortest average
print('Shortest average runtime per director')
display(df.groupby('Director')['Runtime'].mean().sort_values(ascending= True).head())
```

-  *Kenneth Branagh* has the longest average runtime (**242 min**) and *Buster Keaton* the shortest (**45 min**). 

### - Does users ratings match the metascore ratings ?
        - two ratings (IMDB_Rating - Meta_score)

```python
# IMDB rating

top_IMDB = df.sort_values('IMDB_Rating', ascending= False)[['Series_Title', 'Released_Year','No_of_Votes', 'IMDB_Rating']].head(10)
top_IMDB
```

```python
# Meta score

top_Meta = df.sort_values('Meta_score', ascending= False)[['Series_Title', 'Released_Year','No_of_Votes', 'Meta_score']].head(10)
top_Meta
```

```python
# Movies with both scores

both = df[['IMDB_Rating', 'Meta_score']].dropna()
print(f'Movies with both scores: {len(both)}')

# Movie in both top list
print('Movie in both top-list:',set(top_IMDB['Series_Title']) & set(top_Meta['Series_Title']))
```

- The two top-10 lists have **one movie in common** (*The Godfather*) across all 843 movies that have both scores

### - Is there any correlation between numerical columns

```python
df.corr(numeric_only= True).round(2)
```

- Strongest: `No_of_Votes` with `Gross` (0.57) and with `IMDB_Rating` (0.49) — movies that more people watch and vote on earn more and rated higher.
- Almost none: `IMDB_Rating` with `Gross` (0.10) — a high rating by itself does not mean a big earn.
- Negative: `Meta_score` with `Released_Year` (−0.34) — newer movies tend to have lower Metascores.

# Conclusions

**Cleaning**
- `Drishyam` appears twice but they are two different films, so kept both.
- `Poster_Link, Overview` are droped, not needed for analysis.
- `U/A` and `UA` Merged label to `UA`.
- `Released_Year`, `Runtime` and `Gross` converted from text to numbers, and fixed one year (*Apollo 13*, `PG` → 1995).
-  `Gross`, `Meta_score` and `Certificate` missing values were left as `NaN`.


**Findings**
- The highest-grossing movie is *Star Wars: Episode VII – The Force Awakens* (≈ $936.7M).
- Movies released 2000–2010 earned about $73.7M on average 260 movies.
- Average runtimes range from 45 min (Buster Keaton) to 242 min (Kenneth Branagh).

**Limitations:** 
- the data is a top-1000 list, so ratings are restricted to 7.6–9.3, and 16.9% of movies have no `Gross`.
